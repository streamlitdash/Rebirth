# Archive the official Risk Explorer snapshot

Schedule this notebook after the market becomes official. The operation is idempotent, so a completed market date is never overwritten.

```python
risk_cube_project_root = None
```

```python
from pathlib import Path
import importlib
import os
import sys

configured_root = risk_cube_project_root or os.getenv("RISK_CUBE_PROJECT_ROOT")
candidates = (
    [Path(configured_root)] if configured_root else [Path.cwd(), *Path.cwd().parents]
)
project_root = next(
    (
        path.resolve()
        for path in candidates
        if (path / "tools").is_dir() and (path / "core").is_dir()
    ),
    None,
)
if project_root is None:
    raise RuntimeError(
        "Set RISK_CUBE_PROJECT_ROOT or run the job from the project tree"
    )
sys.path.insert(0, str(project_root))

run_from_env = importlib.import_module("tools.s03_archive_official_risk").run_from_env

result = run_from_env()
print(result)
```
